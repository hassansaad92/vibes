```python
import pickle
from pathlib import Path

import pandas as pd

from train_classifier import extract_features

with open('model.pkl', 'rb') as f:
    bundle = pickle.load(f)

model = bundle['model']
feature_cols = bundle['feature_cols']
use_top_freqs = bundle['use_top_freqs']

print('features:', feature_cols)
print('use_top_freqs:', use_top_freqs)
```

```python
CSV_PATH = Path('data/ftest').glob('*.csv').__next__()
print('predicting on:', CSV_PATH)

row = extract_features(CSV_PATH, label=None, use_top_freqs=use_top_freqs)
X = pd.DataFrame([row])[feature_cols].to_numpy()

pred = model.predict(X)[0]
proba = dict(zip(model.classes_, model.predict_proba(X)[0]))
print('prediction:', pred)
print('probabilities:', proba)
```

```python
for split_dir in ['data/ftest', 'data/htest']:
    paths = sorted(Path(split_dir).glob('*.csv'))[:3]
    for p in paths:
        row = extract_features(p, label=None, use_top_freqs=use_top_freqs)
        X = pd.DataFrame([row])[feature_cols].to_numpy()
        pred = model.predict(X)[0]
        print(f'{p.name:20s} -> {pred}')
```
